```python
# Importa moduli
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data import (
    detect_trends,
    get_all_trends,
    get_correlation_matrix
)
from src.visualization import (
    plot_country_timeseries,
    plot_comparison_countries,
    plot_scatter_relationships,
    plot_correlation_heatmap
)
from src.config import DATA_PROCESSED_DIR, NUMERIC_COLUMNS

# Setup plotting
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)

print("Modules loaded successfully!")
```

```python
# Carica il dataset pulito
df = pd.read_csv(DATA_PROCESSED_DIR / "df_clean.csv")

print(f"Dataset loaded: {df.shape}")
print(f"\nCountries: {sorted(df['Country Name'].unique())}")
print(f"Year range: {int(df['Year'].min())} - {int(df['Year'].max())}")
```

## 1. Analisi dei Trend per Indicatore

```python
# Analizza trend per Secondary Completion Rate
print("\n" + "="*80)
print("Secondary Completion Rate - Trend Analysis")
print("="*80)

trends = get_all_trends(df, 'Secondary Completion Rate (%)')

df_trends = pd.DataFrame(trends).sort_values('change_pct', ascending=False)

for _, row in df_trends.iterrows():
    print(f"\n{row['country']}:")
    print(f"  Direction:    {row['direction']}")
    print(f"  Change:       {row['change_pct']:+.1f}% over {row['years']} years")
    print(f"  Start value:  {row['start_value']:.2f}%")
    print(f"  End value:    {row['end_value']:.2f}%")
    print(f"  Slope:        {row['slope']:.4f}% per year")
```

```python
# Analizza trend per Primary Enrolment Rate
print("\n" + "="*80)
print("Primary Enrolment Rate - Trend Analysis")
print("="*80)

trends = get_all_trends(df, 'Primary Enrolment Rate(%)')

df_trends = pd.DataFrame(trends).sort_values('change_pct', ascending=False)

for _, row in df_trends.iterrows():
    print(f"\n{row['country']}:")
    print(f"  Direction:    {row['direction']}")
    print(f"  Change:       {row['change_pct']:+.1f}% over {row['years']} years")
    print(f"  Start value:  {row['start_value']:.2f}%")
    print(f"  End value:    {row['end_value']:.2f}%")
```

```python
# Analizza trend per Education Expenditure
print("\n" + "="*80)
print("Education Expenditure (% of GDP) - Trend Analysis")
print("="*80)

trends = get_all_trends(df, 'Education Expenditure (% of GDP)')

df_trends = pd.DataFrame(trends).sort_values('change_pct', ascending=False)

for _, row in df_trends.iterrows():
    print(f"\n{row['country']}:")
    print(f"  Direction:    {row['direction']}")
    print(f"  Change:       {row['change_pct']:+.1f}% over {row['years']} years")
    print(f"  Start value:  {row['start_value']:.2f}% of GDP")
    print(f"  End value:    {row['end_value']:.2f}% of GDP")
```

## 2. Relazioni tra Educazione e Sviluppo

```python
# Relazione tra Secondary Completion e Life Expectancy
fig = plot_scatter_relationships(
    df,
    x_col='Secondary Completion Rate (%)',
    y_col='Life Expectancy (Years)',
    size_col='GDP per Capita',
)
fig.show()

# Calcola correlazione
corr = df['Secondary Completion Rate (%)'].corr(df['Life Expectancy (Years)'])
print(f"\nCorrelazione: {corr:.3f}")
```

```python
# Relazione tra Secondary Completion e Infant Mortality
fig = plot_scatter_relationships(
    df,
    x_col='Secondary Completion Rate (%)',
    y_col='Infant Mortality Rate (per 1000 live births)',
    color_col='Country Name'
)
fig.show()

# Calcola correlazione (dovrebbe essere negativa)
corr = df['Secondary Completion Rate (%)'].corr(df['Infant Mortality Rate (per 1000 live births)'])
print(f"\nCorrelazione: {corr:.3f}")
print("(Correlazione negativa indica: più alta l'istruzione, più bassa la mortalità infantile)")
```

```python
# Relazione tra Spesa Educativa e GDP pro capite
fig = plot_scatter_relationships(
    df,
    x_col='Education Expenditure (% of GDP)',
    y_col='GDP per Capita',
    color_col='Country Name'
)
fig.show()

# Calcola correlazione
corr = df['Education Expenditure (% of GDP)'].corr(df['GDP per Capita'])
print(f"\nCorrelazione: {corr:.3f}")
```

## 3. Comparazione Paesi Specifici

```python
# Seleziona alcuni paesi interessanti per analisi dettagliata
interesting_countries = ['Brazil', 'India', 'China', 'South Africa']

for country in interesting_countries:
    print(f"\n{'='*80}")
    print(f"Analysis: {country}")
    print(f"{'='*80}")
    
    country_data = df[df['Country Name'] == country]
    
    print(f"\nData coverage: {country_data['Year'].min():.0f} - {country_data['Year'].max():.0f} ({len(country_data)} records)")
    print(f"\nLatest values ({country_data['Year'].max():.0f}):")
    
    latest = country_data[country_data['Year'] == country_data['Year'].max()].iloc[0]
    
    print(f"  Secondary Completion:    {latest['Secondary Completion Rate (%)']:.1f}%")
    print(f"  Primary Enrolment:       {latest['Primary Enrolment Rate(%)']:.1f}%")
    print(f"  Education Expenditure:   {latest['Education Expenditure (% of GDP)']:.2f}% of GDP")
    print(f"  GDP per Capita:          ${latest['GDP per Capita']:,.0f}")
    print(f"  Life Expectancy:         {latest['Life Expectancy (Years)']:.1f} years")
    print(f"  Infant Mortality:        {latest['Infant Mortality Rate (per 1000 live births)']:.1f} per 1000 births")
```

```python
# Confronto dei paesi in developing countries vs developed
developing = ['Brazil', 'India', 'China', 'South Africa', 'Kenya']
developed = ['Italy', 'Germany', 'United States', 'Canada', 'Australia']

fig = plot_comparison_countries(
    df,
    indicator='Secondary Completion Rate (%)',
    countries=developing + developed
)
fig.show()
```

```python
# Confronto aspettativa di vita
fig = plot_comparison_countries(
    df,
    indicator='Life Expectancy (Years)',
    countries=developing + developed
)
fig.show()
```
