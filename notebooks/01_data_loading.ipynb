# Step 1: Data Loading

Questo notebook carica tutti i dataset dalla cartella `data/raw/`.

I file vengono caricati utilizzando percorsi **dinamici** gestiti da `src/config.py`.
Questo significa che il codice funzionerà ovunque venga spostato il progetto.

```python
# SETUP: Configure Python path to find src/ module
# This is necessary because the notebook is in notebooks/ but src/ is at project root
import sys
from pathlib import Path

# Get the project root (parent of the notebooks directory)
notebook_dir = Path.cwd()
project_root = notebook_dir.parent  # Go up one level from notebooks/

# Add project root to Python path so we can import src/
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
    print(f"✓ Added to sys.path: {project_root}")

# Verify the path is configured
print(f"✓ Current working directory: {notebook_dir}")
print(f"✓ Project root: {project_root}")
print(f"✓ src/ module should be at: {project_root}/src")
```

```python
# Importa le funzioni di caricamento dal modulo src.data
from src.data import load_all_data
from src.config import DATA_FILES, COUNTRIES_LIST, MIN_YEAR

print("Modules loaded successfully!")
```

```python
# Verifica che i file siano raggiungibili
print("\nChecking data files...\n")

for key, file_path in DATA_FILES.items():
    exists = file_path.exists()
    status = "✓" if exists else "✗"
    print(f"{status} {key}: {file_path}")

print(f"\nAnalysis parameters:")
print(f"  - Countries: {len(COUNTRIES_LIST)}")
print(f"  - Min year: {MIN_YEAR}")
```

```python
# Carica tutti i dataset e uniscili
df_final = load_all_data()
```

```python
# Visualizza prime righe del dataset finale
print("\nFirst rows of final dataset:")
df_final.head()
```

```python
# Informazioni sul dataset finale
print("\nDataset Info:")
df_final.info()
```

```python
# Verifica dati mancanti prima della pulizia
print("\nMissing values per column:")
print(df_final.isna().sum())
```

```python
# Salva il dataset "grezzo" (prima della pulizia)
import os
from src.config import DATA_PROCESSED_DIR

# Assicura che la directory esista
DATA_PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

output_file = DATA_PROCESSED_DIR / "df_raw_merged.csv"
df_final.to_csv(output_file, index=False)

print(f"✓ Dataset saved to: {output_file}")
print(f"  Shape: {df_final.shape}")
```
